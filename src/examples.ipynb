```python
import json
from backend.api.model.prompt_enhancing_model import ClarifyingUserIntent
from backend.utils.prompts import build_enhanced_output
import httpx
```

```python
from backend.utils.prompts import clarify_user_intent

raw_prompt = "Write a text in 4chan greentext style about the manager of a bottomless pit. The absurder and funnier, the better"
output = clarify_user_intent(raw_prompt, "cloud")
```

```python
structured_output = json.loads(output.json()["output"])
intent = ClarifyingUserIntent(**structured_output)
```

```python
enhanced_prompt = build_enhanced_output(intent)
print(enhanced_prompt)
```
